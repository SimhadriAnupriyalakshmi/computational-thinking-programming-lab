```python
import time
import sys

n = 100000


# List processing
start = time.time()

list_data = [i * i for i in range(n)]

list_time = time.time() - start


# Generator processing
start = time.time()

generator_data = (i * i for i in range(n))

generator_time = time.time() - start


print("List Processing Time:",
      round(list_time, 6), "seconds")

print("Generator Creation Time:",
      round(generator_time, 6), "seconds")

print("List Memory:",
      sys.getsizeof(list_data), "bytes")

print("Generator Memory:",
      sys.getsizeof(generator_data), "bytes")
```

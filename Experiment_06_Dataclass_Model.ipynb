```python
from dataclasses import dataclass


# Traditional class
class Student:

    def __init__(self, name: str, age: int, branch: str):
        self.name = name
        self.age = age
        self.branch = branch


# Dataclass
@dataclass
class StudentData:
    name: str
    age: int
    branch: str


# Objects
student1 = Student(
    "Anupriya",
    21,
    "ECE"
)

student2 = StudentData(
    "Anupriya",
    21,
    "ECE"
)


print("Traditional Class:")
print("Name:", student1.name)
print("Age:", student1.age)
print("Branch:", student1.branch)


print("\nDataclass:")
print(student2)
```

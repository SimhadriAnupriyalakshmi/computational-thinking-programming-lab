```python
def calculate_grade(marks: list[int]) -> str:
    average = sum(marks) / len(marks)

    if average >= 90:
        return "A"
    elif average >= 75:
        return "B"
    elif average >= 60:
        return "C"
    else:
        return "D"

marks = [80, 75, 90]

print("Marks:", marks)
print("Grade:", calculate_grade(marks))
```
